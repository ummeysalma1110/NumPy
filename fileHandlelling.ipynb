practice problem 1
number  of lines , words , character and save tthis to a string files string.txt

```python
with open("./sample_data/sample.txt", "r") as file:
   string = file.readlines()
   print(string)

   #line
   totalLines = len(string)
   print(totalLines)

   #word
   totalWords = 0
   for line in string:
     totalWords += len(line.split())
   print(totalWords)

   #number of character
   #age new line clean korbo

   string = list(map(str.strip, string))
   #space delete korbo
   string = list(map(lambda x : x.replace(" ",""), string))
   numberOfCharacter = list(map(lambda x : len(x), string))
   print(numberOfCharacter)
   totalCharacter = sum(numberOfCharacter)
   print(totalCharacter)

   with open("./sample_data/string.txt", "w") as file:
    file.write(f"totalLines = {totalLines}\n")
    file.write(f"totalWords = {totalWords}\n")
    file.write(f"totalCharacter = {totalCharacter}\n")


```
