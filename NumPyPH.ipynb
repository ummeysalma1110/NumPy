### 1D Array Indexing and Slicing

```python
# Create 1D array
arr1 = np.array([10, 20, 30, 40, 50, 60])

# 1. Indexing (Extracting specific elements)
print("3rd element (Index 2):", arr1[2])
print("Last element (Negative Index -1):", arr1[-1])

# 2. Slicing [start:stop:step]
print("Index 1 to 4 (excluding 4):", arr1[1:4])
print("From beginning to index 3:", arr1[:4])
print("From index 2 to end:", arr1[2:])
print("Skipping one element (Step 2):", arr1[::2])
```

### 2D Array Indexing and Slicing
For 2D arrays, the syntax is `arr[row, column]` or for slicing: `arr[row_start:row_stop, col_start:col_stop]`

```python
# Create 2D array (2 rows, 3 columns)
arr2 = np.array([[10, 20, 30],
                 [40, 50, 60]])

# 1. Indexing [row, col]
print("1st row, 3rd column (Row 0, Col 2):", arr2[0, 2])
print("2nd row, 2nd column (Row 1, Col 1):", arr2[1, 1])

# 2. Selecting full row or column
print("1st row (Row 0):", arr2[0, :])
print("2nd column (Col 1):", arr2[:, 1])

# 3. 2D Slicing
# Extract 1st and 2nd columns from all rows
print("Slicing result:\n", arr2[:, 0:2])
```

### Advanced Indexing and Slicing

```python
arr = np.array([10, 15, 20, 25, 30])

# 1. Boolean Indexing (Filtering data based on condition)
# Get elements greater than 18
condition = arr > 18
print("Condition Mask:", condition)
print("Filtered Elements:", arr[condition])

# 2. Fancy Indexing (Extracting elements using a list of indices)
# Extract elements at index 0, 1, and 4
indices = [0, 1, 4]
print("Elements at indices [0, 1, 4]:", arr[indices])
```

### NumPy Array Iteration
Iterating through 1D and 2D arrays using standard loops and `np.nditer()`.

```python
# 1. Iterating through a 1D Array
print("=== 1D Array Iteration ===")
arr_1d = np.array([10, 20, 30])
for x in arr_1d:
    print(x)

# 2. Iterating through a 2D Array (Row by Row)
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
for row in arr_2d:
    print("Row:", row)

# 3. Iterating through each individual element in a 2D Array
for row in arr_2d:
    for val in row:
        print(val)

# 4. Using np.nditer() for efficient iteration of any dimension
for x in np.nditer(arr_2d):
    print(x)
```

```python
#data load

data = np.genfromtxt('student_scores.csv' ,delimiter=',',skip_header=1)

print(data)
```

```python
# statistics
math_marks = data[::,:1]

print(math_marks.T)
#eikhne .t diye inr chara  baki ggulo remove kora hoise

max_math_marks = np.max(math_marks) #slcing korchi
print(max_math_marks)

min_math_marks = np.min(math_marks)
print(min_math_marks)

average_math_marks = np.mean(math_marks)
print(average_math_marks)

median_math_marks = np.median(math_marks)
print(median_math_marks)

# standard deviation mane egulo ki diverse naki avrg er kacha kachi

std_math_marks = np.std(math_marks)
print(std_math_marks)


```
